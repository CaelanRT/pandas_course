```python
import numpy as np
```

Slicing Arrays - array[start:stop:step]

```python
#1d
product_array = np.array(['fruits', 'vegetables','cereal', 'dairy', 'eggs', 'snacks', 'beverages', 'coffee', 'tea', 'spices'])

print(product_array[1])
print(product_array[-1])
# first 5
product_array[:5]
# starts at index 5, no stop so to end, then skips every 2
product_array[5::2]
```

Slicing 2d arrays - array[row index, column index], array[start:stop:step, start:stop:step]

```python
product_array2D = product_array.reshape(2, 5)
print(product_array2D)
product_array2D[1,2]
# goes through each row, chooses columns from index 2 to the end
product_array2D[:,2:]
product_array2D[1:, :]
```
