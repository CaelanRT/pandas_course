```python
import numpy as np
import pandas as pd
```

```python
my_series = pd.Series([1, np.nan, 2, 3, 4], index=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'] )
my_series
```

```python
my_series + 1
```

```python
# can use fill value for adding a missing value to fill
my_series2 = my_series.add(1, fill_value=0).astype('int')
```

```python
my_series2 /2
```

```python
my_series.add(my_series2, fill_value=0)
```
