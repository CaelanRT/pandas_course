```python
import numpy as np
import pandas as pd
```

```python
# can use equality tests to return an array of boolean operators based on what is present in the series or DF
# ~series.index.isn() inverts your logical boolean values to test for not in!
```

```python
my_series = pd.Series(range(5), index=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'] )
my_series
```

```python
# checks for the values
my_series.loc[~my_series.isin([1,2])]
```

```python
my_series.loc[my_series > 2]
```

```python
# pandas methods as well
my_series.loc[my_series.gt(2)]
```

```python
mask = (my_series.isin([1,2])) | (my_series > 2)
my_series.loc[mask]
```
