```python
import numpy as np

rng = np.random.default_rng(616)

price = (rng.random(10) * 10).round(2)

inventory = rng.integers(0, 100, 10)

product_value = price*inventory
product_value
```

```python
#out of place sort
np.sort(product_value)
```

```python
#inplace sort
product_value.sort()
product_value
```

```python
#reverse array
product_value[::-1]
```
